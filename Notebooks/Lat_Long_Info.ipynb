```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
df = pd.read_csv(r"C:\Users\avanindra Bose\OneDrive\Desktop\Real State Project\Cleaned Datasets\gurgaon_properties_missing_value_imputation.csv")
```

```python
df.head()
```

```python
latlong = pd.read_csv(r"C:\Users\avanindra Bose\OneDrive\Desktop\Real State Project\Cleaned Datasets\latlong.csv")
```

```python
latlong.head()
```

```python
latlong['latitude'] = latlong['coordinates'].str.split(',').str.get(0).str.split('°').str.get(0).astype('float')
```

```python
latlong['longitude'] = latlong['coordinates'].str.split(',').str.get(1).str.split('°').str.get(0).astype('float')
```

```python

latlong.head()
```

```python
new_df = df.merge(latlong, on='sector')
```

```python
new_df.columns
```

```python
group_df = new_df.groupby('sector').mean(numeric_only=True)[['price','price_per_square_feet','built_up_area','latitude','longitude']]
```

```python
group_df[['price','price_per_square_feet','built_up_area']] = group_df[['price','price_per_square_feet','built_up_area']].round(2)
```

```python
group_df
```

```python
fig = px.scatter_map(group_df, lat="latitude", lon="longitude", color="price_per_square_feet", size='built_up_area',
                  color_continuous_scale=px.colors.cyclical.IceFire, zoom=10,
                  text=group_df.index)
fig.show()
```
